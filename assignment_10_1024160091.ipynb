Q1.  Write a unique paragraph (5-6 sentences) about your favorite topic (e.g., sports, technology, food, books, etc.). 1. Convert text to lowercase and remove punctua on using re. 2. Tokenize the text into words and sentences. 3. Split using split() and word_tokenize() and compare how Python split and NLTK’s word_tokenize() differ. 4. Remove stopwords (using NLTK's stopwords list). 5. Display word frequency distribu on (excluding stopwords).

```python
import re
import string
import nltk

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

# Download required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

# Unique paragraph: 6 sentences
text = """
Artificial intelligence is one of my favorite technologies.
It allows computers to learn from data and make intelligent decisions.
AI is widely used in healthcare, education, banking, and transportation.
Machine learning helps systems improve their performance over time.
AI can also automate repetitive tasks and save human effort.
I believe artificial intelligence will play an important role in the future.
"""

print("Original Text:\n", text)

# 1. Convert text to lowercase and remove punctuation using re
clean_text = text.lower()
clean_text = re.sub(r'[^\w\s.]', '', clean_text)

print("\nLowercase Text with Punctuation Removed:")
print(clean_text)

# 2. Tokenize into words and sentences
words = word_tokenize(clean_text)
sentences = sent_tokenize(text)

print("\nWord Tokens:")
print(words)

print("\nSentence Tokens:")
for sentence in sentences:
    print(sentence)

# 3. Compare split() and word_tokenize()
python_split = clean_text.split()
nltk_tokens = word_tokenize(clean_text)

print("\nPython split() Output:")
print(python_split)

print("\nNLTK word_tokenize() Output:")
print(nltk_tokens)

print("\nDifference:")
print("split() separates text mainly using spaces.")
print("word_tokenize() performs more detailed tokenization and handles punctuation separately.")

# 4. Remove stopwords
stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in words
    if word not in stop_words and word.isalpha()
]

print("\nWords after Removing Stopwords:")
print(filtered_words)

# 5. Word frequency distribution excluding stopwords
frequency = FreqDist(filtered_words)

print("\nWord Frequency Distribution:")
for word, count in frequency.items():
    print(word, ":", count)
```

Q2. Using the same paragraph from Q1: 1. Extract all words with only alphabets using re.findall() 2. Remove stop words using NLTK’s stopword list 3. Perform stemming with PorterStemmer 4. Perform lemma za on with WordNetLemma zer 5. Compare the stemmed and lemma zed outputs and explain when you’d prefer one over the other.

```python
import re
import nltk

from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download required resources
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

text = """
Artificial intelligence is one of my favorite technologies.
It allows computers to learn from data and make intelligent decisions.
AI is widely used in healthcare, education, banking, and transportation.
Machine learning helps systems improve their performance over time.
AI can also automate repetitive tasks and save human effort.
I believe artificial intelligence will play an important role in the future.
"""

# 1. Extract only alphabetic words using re.findall()
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

print("Alphabetic Words:")
print(words)

# 2. Remove stopwords
stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in words
    if word not in stop_words
]

print("\nAfter Removing Stopwords:")
print(filtered_words)

# 3. Stemming using PorterStemmer
stemmer = PorterStemmer()

stemmed_words = [
    stemmer.stem(word)
    for word in filtered_words
]

print("\nStemmed Words:")
print(stemmed_words)

# 4. Lemmatization using WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in filtered_words
]

print("\nLemmatized Words:")
print(lemmatized_words)

# 5. Comparison
print("\nComparison:")
print("Stemming removes prefixes or suffixes and may produce words that are not valid English words.")
print("Lemmatization converts words into meaningful dictionary base forms.")
print("Stemming is useful when speed is important.")
print("Lemmatization is preferred when correct linguistic meaning is important.")
```

Q3. Choose 3 short texts of your own (e.g., different news headlines, product reviews). 1. Use CountVectorizer to generate the Bag of Words representa on. 2. Use TfidfVectorizer to compute TF-IDF scores. 3. Print and interpret the top 3 keywords from each text using TF-IDF.

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
import numpy as np

# Three short texts
texts = [
    "Artificial intelligence is transforming healthcare using smart diagnostic systems.",

    "Electric vehicles are becoming popular because they reduce pollution and fuel consumption.",

    "Cloud computing provides scalable storage and computing resources over the internet."
]



count_vectorizer = CountVectorizer(stop_words='english')

bow_matrix = count_vectorizer.fit_transform(texts)

print("Bag of Words Vocabulary:")
print(count_vectorizer.get_feature_names_out())

print("\nBag of Words Matrix:")
print(bow_matrix.toarray())



tfidf_vectorizer = TfidfVectorizer(stop_words='english')

tfidf_matrix = tfidf_vectorizer.fit_transform(texts)

feature_names = tfidf_vectorizer.get_feature_names_out()

print("\nTF-IDF Vocabulary:")
print(feature_names)

print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())



print("\nTop 3 Keywords from Each Text:")

for i in range(len(texts)):

    scores = tfidf_matrix[i].toarray()[0]

    top_indices = scores.argsort()[-3:][::-1]

    top_keywords = [
        (feature_names[index], scores[index])
        for index in top_indices
    ]

    print("\nText", i + 1, ":")
    print(texts[i])

    print("Top 3 Keywords:")

    for word, score in top_keywords:
        print(word, ":", round(score, 4))
```

Q4. Write 2 short texts (4–6 lines each) describing two different technologies (e.g., AI vs Blockchain). 1. Preprocess and tokenize both texts. 2. Calculate: a. Jaccard Similarity using sets b. Cosine Similarity using TfidfVectorizer + cosine_similarity() c. Analyze which similarity metric gives be er insights in your case.

```python
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


# Two technology texts
text1 = """
Artificial intelligence enables machines to perform tasks that normally require human intelligence.
AI systems can learn from data and identify useful patterns.
It is widely used in healthcare, banking, transportation, and education.
Machine learning is an important part of artificial intelligence.
AI can automate tasks and improve decision making.
"""

text2 = """
Blockchain is a decentralized technology used to store digital transactions securely.
It stores information in blocks that are connected using cryptography.
Blockchain is widely used in cryptocurrencies, banking, and supply chain systems.
It improves transparency and reduces the need for central authorities.
Blockchain technology can provide secure and reliable digital records.
"""


# --------------------------------------------------
# 1. Preprocess and tokenize
# --------------------------------------------------

def preprocess(text):

    # Convert to lowercase
    text = text.lower()

    # Extract alphabetic words only
    words = re.findall(r'\b[a-zA-Z]+\b', text)

    return words


tokens1 = preprocess(text1)
tokens2 = preprocess(text2)

print("Tokens of Text 1:")
print(tokens1)

print("\nTokens of Text 2:")
print(tokens2)


# --------------------------------------------------
# 2a. Jaccard Similarity
# --------------------------------------------------

set1 = set(tokens1)
set2 = set(tokens2)

intersection = set1.intersection(set2)
union = set1.union(set2)

jaccard_similarity = len(intersection) / len(union)

print("\nCommon Words:")
print(intersection)

print("\nJaccard Similarity:")
print(round(jaccard_similarity, 4))




tfidf = TfidfVectorizer(stop_words='english')

tfidf_matrix = tfidf.fit_transform([text1, text2])

cosine_sim = cosine_similarity(
    tfidf_matrix[0:1],
    tfidf_matrix[1:2]
)

print("\nCosine Similarity:")
print(round(cosine_sim[0][0], 4))


print("\nAnalysis:")

if cosine_sim[0][0] > jaccard_similarity:
    print("Cosine Similarity gives better insight in this case.")
else:
    print("Jaccard Similarity gives better insight in this case.")

print("""
Jaccard Similarity only checks the overlap of unique words between the two texts.
It does not consider how important a word is in a document.

Cosine Similarity using TF-IDF is more informative because it considers
the importance and frequency of words in both texts.

Therefore, Cosine Similarity gives better insight for comparing these
technology descriptions because it captures the overall similarity
between the texts more effectively.
""")
```

Q5. Write a short review for a product or service. 1. Use TextBlob or VADER to find polarity & subjec vity for each review. 2. Classify reviews into Posi ve / Nega ve / Neutral. 3. Create a word cloud using the wordcloud library for all posi ve reviews.

```python
from textblob import TextBlob
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# Short product/service reviews
reviews = [
    "The smartphone is excellent and the battery life is amazing.",
    "The camera quality is very poor and the phone becomes slow.",
    "The design is simple and the performance is average.",
    "I really love this product because it is fast and reliable.",
    "The customer service was terrible and very disappointing."
]

positive_reviews = []

# Analyze each review
for i, review in enumerate(reviews, start=1):

    blob = TextBlob(review)

    polarity = blob.sentiment.polarity
    subjectivity = blob.sentiment.subjectivity

    # Classify sentiment
    if polarity > 0:
        sentiment = "Positive"
        positive_reviews.append(review)

    elif polarity < 0:
        sentiment = "Negative"

    else:
        sentiment = "Neutral"

    print("Review", i, ":", review)
    print("Polarity:", round(polarity, 3))
    print("Subjectivity:", round(subjectivity, 3))
    print("Sentiment:", sentiment)
    print("-" * 50)


# Create Word Cloud for positive reviews
positive_text = " ".join(positive_reviews)

wordcloud = WordCloud(
    width=800,
    height=400,
    background_color="white"
).generate(positive_text)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("Word Cloud of Positive Reviews")
plt.show()
```

Q6. Choose your own paragraph (~100 words) as training data. 1. Tokenize text using Tokenizer() from keras.preprocessing.text 2. Create input sequences and build a simple LSTM or Dense model 3. Train the model and generate 2–3 new lines of text star ng from any seed word you provide.

```python
import numpy as np
import tensorflow as tf

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense


# --------------------------------------------------
# Training paragraph (~100 words)
# --------------------------------------------------

text = """
Artificial intelligence is changing the way people use technology.
It helps computers learn from data and make useful decisions.
Machine learning is an important part of artificial intelligence.
It is used in healthcare, education, banking, transportation and many other fields.
Smart systems can recognize images, understand speech and analyze large amounts of information.
Artificial intelligence can also automate repetitive tasks and improve productivity.
In the future, intelligent machines may become more helpful in daily life.
However, these technologies should be developed carefully and used responsibly for the benefit of society.
"""


# --------------------------------------------------
# 1. Tokenize text using Tokenizer()
# --------------------------------------------------

tokenizer = Tokenizer()

tokenizer.fit_on_texts([text])

total_words = len(tokenizer.word_index) + 1

print("Total Words:", total_words)

print("\nWord Index:")
print(tokenizer.word_index)


# --------------------------------------------------
# 2. Create input sequences
# --------------------------------------------------

input_sequences = []

for line in text.split("."):

    token_list = tokenizer.texts_to_sequences([line])[0]

    for i in range(1, len(token_list)):

        n_gram_sequence = token_list[:i + 1]

        input_sequences.append(n_gram_sequence)


# Find maximum sequence length
max_sequence_len = max(
    [len(sequence) for sequence in input_sequences]
)

print("\nMaximum Sequence Length:", max_sequence_len)


# Pad sequences
input_sequences = np.array(
    pad_sequences(
        input_sequences,
        maxlen=max_sequence_len,
        padding='pre'
    )
)


# Split into predictors and labels
X = input_sequences[:, :-1]

labels = input_sequences[:, -1]


# Convert labels into categorical format
y = tf.keras.utils.to_categorical(
    labels,
    num_classes=total_words
)


print("\nShape of X:", X.shape)
print("Shape of y:", y.shape)


# --------------------------------------------------
# Build simple LSTM model
# --------------------------------------------------

model = Sequential()

model.add(
    Embedding(
        total_words,
        50,
        input_length=max_sequence_len - 1
    )
)

model.add(
    LSTM(100)
)

model.add(
    Dense(
        total_words,
        activation='softmax'
    )
)


model.compile(
    loss='categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)


model.summary()


# --------------------------------------------------
# 3. Train the model
# --------------------------------------------------

model.fit(
    X,
    y,
    epochs=100,
    verbose=1
)


# --------------------------------------------------
# Generate new text
# --------------------------------------------------

seed_text = "artificial intelligence"

next_words = 25


for _ in range(next_words):

    token_list = tokenizer.texts_to_sequences([seed_text])[0]

    token_list = pad_sequences(
        [token_list],
        maxlen=max_sequence_len - 1,
        padding='pre'
    )

    predicted = np.argmax(
        model.predict(token_list, verbose=0),
        axis=-1
    )[0]


    output_word = ""

    for word, index in tokenizer.word_index.items():

        if index == predicted:

            output_word = word
            break


    seed_text += " " + output_word


print("\nGenerated Text:")
print(seed_text)
```
