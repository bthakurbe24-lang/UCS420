```python
# Q1. List Operations

L = [10, 20, 30, 40, 50, 60, 70, 80]

# i. Add 200 and 300 to L
L.append(200)
L.append(300)
print("After adding 200 and 300:", L)

# ii. Remove 10 and 30 from L
L.remove(10)
L.remove(30)
print("After removing 10 and 30:", L)

# iii. Sort L in ascending order
L.sort()
print("Ascending order:", L)

# iv. Sort L in descending order
L.sort(reverse=True)
print("Descending order:", L)
```

```python
# Q2. Tuple Operations

scores = (45, 89.5, 76.45, 89, 75.6, 89.4, 89, 92, 58, 45)

# i. Identify the highest score and its index
highest = max(scores)
index = scores.index(highest)

print("Highest score:", highest)
print("Index of highest score:", index)

# ii. Find the lowest score and count how many times it appears
lowest = min(scores)
count_lowest = scores.count(lowest)

print("Lowest score:", lowest)
print("Number of times lowest score appears:", count_lowest)

# iii. Reverse the tuple and return it as a list
reversed_list = list(scores[::-1])

print("Reversed tuple as list:", reversed_list)

# iv. Check whether 76 is present
if 76 in scores:
    print("76 is present in the tuple")
else:
    print("76 is not present in the tuple")
```

```python
# Q3. Generate 100 random numbers between 100 and 900

import random

numbers = []

# Generate 100 random numbers
for i in range(100):
    numbers.append(random.randint(100, 900))

# Find odd numbers
odd_numbers = []

for num in numbers:
    if num % 2 != 0:
        odd_numbers.append(num)

# Find even numbers
even_numbers = []

for num in numbers:
    if num % 2 == 0:
        even_numbers.append(num)

# Function to check prime number
def is_prime(num):
    if num < 2:
        return False

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False

    return True


# Find prime numbers
prime_numbers = []

for num in numbers:
    if is_prime(num):
        prime_numbers.append(num)

print("All 100 numbers:")
print(numbers)

print("\nOdd numbers:")
print(odd_numbers)

print("\nEven numbers:")
print(even_numbers)

print("\nPrime numbers:")
print(prime_numbers)
```

```python
# Q4. Set Operations

A = {34, 56, 78, 90}
B = {78, 45, 90, 23}

# i. Union - unique scores achieved by both teams
print("Union:", A.union(B))

# ii. Intersection - scores common to both teams
print("Intersection:", A.intersection(B))

# iii. Symmetric Difference - scores exclusive to each team
print("Symmetric Difference:", A.symmetric_difference(B))

# iv. Check if B is a subset of A
if B.issubset(A):
    print("B is a subset of A")
else:
    print("B is not a subset of A")

# v. Remove a specific score X entered by the user
X = int(input("Enter the score to remove from A: "))

if X in A:
    A.remove(X)
    print("Updated set A:", A)
else:
    print("Score is not present in set A")
```
