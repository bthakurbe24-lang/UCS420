assignment 5.3 cc

```python
n = int(input("enter the number:"))
sum = 0
for i in range (2,n+1):
  prime = True
  for j in range(2,i):
    if i % j == 0:
      prime = False
      break
      if prime:
        sum = sum + 1
        print("sum of prime number:",sum)
```
